### Now to select more comments that have subreddit values Bitcoin from the original dataset filter  the comments that have subreddit value Bitcoin

```python
# Create a new dataframe named bitcoin_data that contains only the rows from original_data where the "subreddit" column is equal to "Bitcoin"
bitcoin_data = original_data[original_data["subreddit"] == "Bitcoin"].copy()

# Save this data to a new CSV file named "bitcoin_only_data.csv" in the same directory as the original dataset
output_file_path = r"/Users/mariyajose/Documents/Sarcasm/dataset/bitcoin_only_data.csv"

# Save the filterd data inside the dataframe bitcoin_data to the output file path as a CSV file without the index column
bitcoin_data.to_csv(output_file_path, index=False)

print("Dataframe 'bitcoin_data' has been saved to the following path:", output_file_path)
print("The number of rows in the 'bitcoin_data' dataframe is:", len(bitcoin_data))
```

### After reading through the comments inside the bitcoin_data some of the comments inside the parent comment column appeared confusing and with bery short number of words. Hence to make the process much more meaningfull the study only select the comments that has the word bitcoin or other words indicating bitcoin in it.The parent comment coulmn is focused for filtering, because it show the reason or background of the comment that is being tested for sarcastic or non sarcastic comments.

```python
# Define the column that needs to be filtered for the keywords
column_under_consideration = "parent_comment"

# Check if the column exists in the bitcoin_data used for the study 
if column_under_consideration in bitcoin_data.columns:
    
    # Define the regex pattern to match the keywords (case-insensitive)
    searching_pattern = r"\b(BTC|BT|BU|ELI5)\b|bit\s*coin"

    # Create a boolean mask for rows that contain the keywords in the specified column
    boolean_keyword_mask = bitcoin_data[column_under_consideration].str.contains(
        searching_pattern, flags=re.IGNORECASE, na=False
    )

    # Use the boolean mask to filter the bitcoin_data DataFrame
    final_bitcoin_data = bitcoin_data[boolean_keyword_mask]

    # Define the path  to save the final filtered DataFrame with keyword matches
    final_output_path = r"/Users/mariyajose/Documents/Sarcasm/dataset/final_bitcoin_data.csv"

    # Save the final filtered DataFrame to a CSV file without the index column
    final_bitcoin_data.to_csv(final_output_path, index=False)

    print("Final filtered DataFrame 'final_bitcoin_data' has been saved to the following path:", final_output_path)
    print("The number of rows in the 'final_bitcoin_data' dataframe is:", len(final_bitcoin_data))
    print("Data saved successfully to:", final_output_path)
else:
    print("Column not found in the dataset.")
```
