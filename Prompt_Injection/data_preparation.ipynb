```python
# import the necessary packages
import random
import pandas as pd
```

```python
# 1. Read your existing CSV file
# Replace 'your_input_file.csv' with the actual path to your file
data = pd.read_csv("/Users/mariyajose/Documents/Sarcasm_Final/Exploratory_Data_Analysis/final_bitcoin_data.csv")

# 2. Define the prompt lists
prompts_for_injecting_to_sarcastic_comments = [
    "not being funny here",
    "i mean it",
    "not kidding",
    "my honest openinon",
    "jokes aside",
    "not being playfull here",
    "for real though",
    "frankly speaking",
    "to be frank",
    "pure truth",
]

prompts_for_injecting_to_non_sarcastic_comments = [
    "i was just jocking",
    "haha, my bad attempt at humour ",
    "….ironically speaking",
    "i didnt mean it literally",
    "by the way…that was a joke",
    "i was kidding",
    "dont take it literally",
    "obviously not",
    "Just poking fun",
    "irony at its finest",
]


# 3. Define the injection function
def prompt_injection(row):
    comment = str(row["comment"])
    label = row["label"]

    if label == 1:
        prompt = random.choice(prompts_for_injecting_to_sarcastic_comments)
    elif label == 0:
        prompt = random.choice(prompts_for_injecting_to_non_sarcastic_comments)
    else:
        return comment

    # Randomly select placement: beginning or end
    position = random.choice(["beginning", "end"])

    if position == "beginning":
        return f"{prompt} {comment}"
    else:
        return f"{comment} {prompt}"


# 4. Modify the 'comment' column
data["comment"] = data.apply(prompt_injection, axis=1)

# 5. Save the updated DataFrame to a new CSV file
# Change 'output_file.csv' to whatever name you prefer
data.to_csv("/Users/mariyajose/Documents/Sarcasm_Final/Prompt_Injection/prompt_injected_data.csv", index=False)

print("Prompt injection completed and saved to 'prompt_injected_data.csv'.")
```
